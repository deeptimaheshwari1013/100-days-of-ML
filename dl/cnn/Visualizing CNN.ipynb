```python
from keras.applications.vgg16 import VGG16
# load the model
model1 = VGG16()

model1.summary()
```

```python
from keras.utils import plot_model

plot_model(model1)
```

```python
for i in range(len(model1.layers)):
  # check for convolutional layer
  if 'conv' not in model1.layers[i].name :
    continue

  # get filter weights
  filters, biases = model1.layers[i].get_weights()
  print("layer number",i,model1.layers[i].name, filters.shape)
```

```python
# retrieve weights from the second hidden layer
filters, bias = model1.layers[1].get_weights()
```

```python
# normalize filter values to 0-1 so we can visualize them
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as plt


n_filters = 6
ix = 1
fig = plt.figure(figsize=(15,10))
for i in range(n_filters):
  # get the filters :
  f= filters[:,:,:,i]

  for j in range(3):

    # subplot for 6 filters and 3 channels
    plt.subplot(n_filters,3,ix)
    plt.imshow(f[:,:,j],cmap='gray')
    ix += 1

# plot the filters
plt.show()
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.vgg16 import preprocess_input
from numpy import expand_dims

# Create a new model to extract features
model1 = Model(
    inputs=model1.inputs,
    outputs=model1.layers[1].output
)

# Load image
image = load_img('/content/img.jpeg', target_size=(224, 224))

# Convert image to array
image = img_to_array(image)

# Expand dimensions
image = expand_dims(image, axis=0)

# Preprocess image
image = preprocess_input(image)
```

```python
# Extract features
features = model1.predict(image)

fig = plt.figure(figsize = (20,15))
for i in range(1, features.shape[3]+1):

  plt.subplot(8,8,i)
  plt.imshow(features[0,:,:,i-1], cmap='gray')

plt.show()
```

```python
model2 = VGG16()

layer_index = [2,5,9,13,17]
outputs = [model2.layers[i].output for i in layer_index]

model3 = Model(inputs = model2.inputs, outputs = outputs)
```

```python
feature_map = model3.predict(image)

for i, fmap in zip(layer_index, feature_map):
  fig = plt.figure(figsize=(20,15))
  fig.suptitle("layer_{}".format(i), fontsize = 20)
  for i in range(1, features.shape[3]+1):
    plt.subplot(8,8,i)
    plt.imshow(fmap[0,:,:,i-1],cmap='gray')

  plt.show()
```
